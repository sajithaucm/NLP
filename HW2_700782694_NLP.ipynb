```python
#Q5. Evaluation Metrics from a Multi-Class Confusion Matrix
import numpy as np

# Confusion matrix: rows = system, cols = gold
cm = np.array([
    [5, 10, 5],   # System: Cat
    [15, 20, 10], # System: Dog
    [0, 15, 10]   # System: Rabbit
])

classes = ["Cat", "Dog", "Rabbit"]

def per_class_metrics(cm):
    num_classes = cm.shape[0]
    precisions = []
    recalls = []

    for i in range(num_classes):
        tp = cm[i, i]
        fp = cm[i, :].sum() - tp
        fn = cm[:, i].sum() - tp

        precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

        precisions.append(precision)
        recalls.append(recall)

        print(f"{classes[i]}: Precision = {precision:.4f}, Recall = {recall:.4f}")

    return np.array(precisions), np.array(recalls)

precisions, recalls = per_class_metrics(cm)

# Macro-averaged
macro_precision = precisions.mean()
macro_recall = recalls.mean()

# Micro-averaged
tp_total = np.trace(cm)
pred_total = cm.sum()
gold_total = cm.sum()

micro_precision = tp_total / pred_total
micro_recall = tp_total / gold_total

print(f"\nMacro Precision = {macro_precision:.4f}")
print(f"Macro Recall    = {macro_recall:.4f}")
print(f"Micro Precision = {micro_precision:.4f}")
print(f"Micro Recall    = {micro_recall:.4f}")
```

```python
# Programming: Bigram Language Model Implementation (based on “Activity: I love NLP corpus” slide)
from collections import defaultdict

# ---------------------------------------------------------
# 1. Training corpus
# ---------------------------------------------------------
corpus = [
    "<s> I love NLP </s>",
    "<s> I love deep learning </s>",
    "<s> deep learning is fun </s>"
]

# ---------------------------------------------------------
# 2. Tokenize sentences
# ---------------------------------------------------------
tokenized = [sentence.split() for sentence in corpus]

# ---------------------------------------------------------
# 3. Compute unigram and bigram counts
# ---------------------------------------------------------
unigram_counts = defaultdict(int)
bigram_counts = defaultdict(lambda: defaultdict(int))

for sentence in tokenized:
    for i in range(len(sentence)):
        unigram_counts[sentence[i]] += 1
        if i < len(sentence) - 1:
            w1 = sentence[i]
            w2 = sentence[i+1]
            bigram_counts[w1][w2] += 1

# ---------------------------------------------------------
# 4. Compute bigram probabilities (MLE)
# ---------------------------------------------------------
bigram_probs = defaultdict(dict)

for w1 in bigram_counts:
    total = sum(bigram_counts[w1].values())
    for w2 in bigram_counts[w1]:
        bigram_probs[w1][w2] = bigram_counts[w1][w2] / total

# ---------------------------------------------------------
# 5. Function to compute probability of a sentence
# ---------------------------------------------------------
def sentence_probability(sentence):
    words = sentence.split()
    prob = 1.0
    for i in range(len(words) - 1):
        w1, w2 = words[i], words[i+1]
        if w2 in bigram_probs[w1]:
            prob *= bigram_probs[w1][w2]
        else:
            prob = 0.0
            break
    return prob

# ---------------------------------------------------------
# 6. Test sentences
# ---------------------------------------------------------
S1 = "<s> I love NLP </s>"
S2 = "<s> I love deep learning </s>"

p1 = sentence_probability(S1)
p2 = sentence_probability(S2)

print("P(S1) =", p1)
print("P(S2) =", p2)

# ---------------------------------------------------------
# 7. Which sentence is preferred?
# ---------------------------------------------------------
if p1 > p2:
    print("Model prefers S1 because it has higher probability.")
else:
    print("Model prefers S2 because it has higher probability.")
```
