```python
# ============================================================
# CS5760 Natural Language Processing - Home Assignment 4 (Part II)
# Student: Sajitha Konda | ID: 700782694 | Fall 2026, UCM
# ============================================================

# ---------- Setup: install libraries and download data ----------
!pip install -q nltk spacy
!python -m spacy download en_core_web_sm -q

import nltk
nltk.download('punkt', quiet=True)       # tokenizer models
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)   # English stopword list

import numpy as np
import spacy
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer


# ============================================================
# Q1: NLP Preprocessing Pipeline
# ============================================================
def preprocess(sentence):
    # Step 1: split the sentence into words and punctuation marks
    tokens = word_tokenize(sentence)
    print("Original Tokens:", tokens)

    # Step 2: remove common English stopwords ("are", "in", "and", ...)
    # Lowercase before comparing so "The" and "the" are both caught;
    # punctuation is also dropped since it carries no meaning here
    stop_words = set(stopwords.words("english"))
    filtered = [t for t in tokens if t.lower() not in stop_words and t.isalnum()]
    print("Tokens Without Stopwords:", filtered)

    # Step 3: reduce each word to its root form with the Porter stemmer
    stemmer = PorterStemmer()
    stemmed = [stemmer.stem(t) for t in filtered]
    print("Stemmed Words:", stemmed)
    return stemmed

print("=" * 60)
print("Q1: NLP Preprocessing Pipeline")
print("=" * 60)
preprocess("NLP techniques are used in virtual assistants like Alexa and Siri.")


# ============================================================
# Q2: Named Entity Recognition with spaCy
# ============================================================
print("\n" + "=" * 60)
print("Q2: Named Entity Recognition with spaCy")
print("=" * 60)

# Load spaCy's small pretrained English pipeline (includes NER)
nlp = spacy.load("en_core_web_sm")

sentence = ("LeBron James as the NBA player of the United States in Laker "
            "and won the NBA Final MVP in 2020.")

# Process the text; spaCy tokenizes it and tags named entities
doc = nlp(sentence)

# Print each entity's text, label, and start/end character positions
for ent in doc.ents:
    print(f"Text: {ent.text:<20} Label: {ent.label_:<8} "
          f"Start: {ent.start_char:<4} End: {ent.end_char}")


# ============================================================
# Q3: Scaled Dot-Product Attention
# ============================================================
def softmax(x):
    # Subtract the row max before exponentiating for numerical stability
    e = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return e / np.sum(e, axis=-1, keepdims=True)

def scaled_dot_product_attention(Q, K, V):
    d = K.shape[-1]                 # key dimension
    scores = Q @ K.T                # 1) dot product of Q and K transpose
    scaled = scores / np.sqrt(d)    # 2) scale by sqrt(d)
    weights = softmax(scaled)       # 3) softmax -> attention weights
    output = weights @ V            # 4) weighted sum of the values
    return weights, output

print("\n" + "=" * 60)
print("Q3: Scaled Dot-Product Attention")
print("=" * 60)

Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

weights, output = scaled_dot_product_attention(Q, K, V)
np.set_printoptions(precision=4, suppress=True)
print("Attention Weights:\n", weights)
print("Output:\n", output)
```
